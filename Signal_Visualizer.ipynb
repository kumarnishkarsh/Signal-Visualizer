```python
import numpy as np
import matplotlib.pyplot as plt

def signal_generation(signal_type, frequency, amplitude, t):
    if signal_type == "sine":
        return amplitude * np.sin(2 * np.pi * frequency * t)
    elif signal_type == "square":
        return amplitude * np.sign(np.sin(2 * np.pi * frequency * t))
    elif signal_type == "triangle":
        return amplitude * (2 * np.abs(2 * (t * frequency - np.floor(t * frequency + 0.5))) - 1)
    else:
        return None

def splot(signal, signal_type, frequency, amplitude, t):
    plt.figure(figsize=(10, 4))
    plt.plot(t, signal, color="royalblue", linewidth=2)
    plt.title(f"{signal_type.capitalize()} Wave | Frequency: {frequency}Hz | Amplitude: {amplitude}", fontsize=14)
    plt.xlabel("Time (s)", fontsize=12)
    plt.ylabel("Amplitude", fontsize=12)
    plt.grid(True, linestyle="--", alpha=0.7)
    plt.axhline(0, color="black", linewidth=0.8)
    plt.tight_layout()
    plt.show()

t = np.linspace(0, 1, 1000)

signal_type = input("Enter signal type (sine / square / triangle): ").strip().lower()
frequency = float(input("Enter frequency (e.g. 5): "))
amplitude = float(input("Enter amplitude (e.g. 2): "))

signal = signal_generation(signal_type, frequency, amplitude, t)

if signal is not None:
    splot(signal, signal_type, frequency, amplitude, t)
else:
    print("Invalid signal type. Please enter sine, square, or triangle.")
```
